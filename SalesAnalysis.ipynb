## Task 1: Identifying top performing states

```python
unique_states = df["State"].nunique()
print(f"Number of unique states: {unique_states}")
```

```python
# Group by state and sum the sales
state_revenue = df.groupby('State')['Sales'].sum()
print(state_revenue)
```

```python
print(f"Average sales amount: ${df['Sales'].mean():,.2f}")
print(f"Median units sold: {df['Unit'].median()}")
print(f"Most common state: {df['State'].mode()[0]}")  # [0] gets first result
print(f"Most popular demographic: {df['Group'].mode()[0]}")

```

```python
# Calculate quartiles for the state revenues
import numpy as np

# Get the revenue values
revenues = state_revenue.values

# Calculate quartiles
q1 = np.percentile(revenues, 25)
q3 = np.percentile(revenues, 75)

print(f"1st Quartile (Q1): ${q1:,.0f}")
print(f"3rd Quartile (Q3): ${q3:,.0f}")

# Classify states by performance
print("\n--- STATE PERFORMANCE CLASSIFICATION ---")
print("🔴 BOTTOM PERFORMERS (1st Quartile - Need Sales Programs):")
for state, revenue in state_revenue.items():
    if revenue <= q1:
        print(f"   {state}: ${revenue:,.0f}")

print("\n🟢 TOP PERFORMERS (4th Quartile - Success Stories):")
for state, revenue in state_revenue.items():
    if revenue >= q3:
        print(f"   {state}: ${revenue:,.0f}")

print("\n🟡 MIDDLE PERFORMERS (2nd & 3rd Quartiles):")
for state, revenue in state_revenue.items():
    if q1 < revenue < q3:
        print(f"   {state}: ${revenue:,.0f}")
```

## Task 2: Develop Sales programs for lower revenue states

Sales analysis of top perfomers

```python
state_revenue = df.groupby('State')['Sales'].sum()
print("State revenues:")
print(state_revenue)
```

```python
top_data = df[df['State'].isin([' NSW', ' VIC'])]
time_revenue = top_data.groupby('Time')['Sales'].sum().sort_values(ascending=False)
print("Revenue by Time of Day:")
print(time_revenue)

# Convert to percentages
total_revenue = time_revenue.sum()
print("\nAs Percentages:")
for time_period, revenue in time_revenue.items():
    percentage = (revenue / total_revenue) * 100
    print(f"{time_period}: {percentage:.1f}% of revenue")
print("\n CUSTOMER GROUP ANALYSIS FOR TOP PERFORMERS")
group_revenue = top_data.groupby('Group')['Sales'].sum().sort_values(ascending=False)
print("Revenue by Customer Group:")
print(group_revenue)

# Convert to percentages
total_revenue_groups = group_revenue.sum()
print("\nAs Percentages:")
for group, revenue in group_revenue.items():
    percentage = (revenue / total_revenue_groups) * 100
    print(f"{group}: {percentage:.1f}% of revenue")
```

Sales analysis of bottom perfomers

```python
# Analyze NT and WA patterns
bottom_data = df[df['State'].isin([' NT', ' WA'])]

print("BOTTOM PERFORMERS (NT + WA) ANALYSIS")
print(f"Total bottom performer revenue: ${bottom_data['Sales'].sum():,}")

print("\n TIME ANALYSIS FOR BOTTOM PERFORMERS")
time_revenue_bottom = bottom_data.groupby('Time')['Sales'].sum().sort_values(ascending=False)
print("Revenue by Time of Day:")
print(time_revenue_bottom)

# Convert to percentages
total_revenue_bottom = time_revenue_bottom.sum()
print("\nAs Percentages:")
for time_period, revenue in time_revenue_bottom.items():
    percentage = (revenue / total_revenue_bottom) * 100
    print(f"{time_period}: {percentage:.1f}% of revenue")

print("\n CUSTOMER GROUP ANALYSIS FOR BOTTOM PERFORMERS")
group_revenue_bottom = bottom_data.groupby('Group')['Sales'].sum().sort_values(ascending=False)
print("Revenue by Customer Group:")
print(group_revenue_bottom)

# Convert to percentages
total_revenue_groups_bottom = group_revenue_bottom.sum()
print("\nAs Percentages:")
for group, revenue in group_revenue_bottom.items():
    percentage = (revenue / total_revenue_groups_bottom) * 100
    print(f"{group}: {percentage:.1f}% of revenue")
```

```python
# Average sales per transaction
top_avg = df[df['State'].isin([' NSW', ' VIC'])]['Sales'].mean()
bottom_avg = df[df['State'].isin([' NT', ' WA'])]['Sales'].mean()
print(f"Top performers avg transaction: ${top_avg:,.0f}")
print(f"Bottom performers avg transaction: ${bottom_avg:,.0f}")
```

Top and Bottom performers show a balanced distribution of sales in terms of times when sales are done as well as the demographics that buy. The issue is then not that the bottom have a specialized approach for instance, selling at a specific time of day, to a specific demographic.  

Further Analysis to determine what is driving higher sales for top performing states:

```python
# Check average units per transaction
top_units = df[df['State'].isin([' NSW', ' VIC'])]['Unit'].mean()
bottom_units = df[df['State'].isin([' NT', ' WA'])]['Unit'].mean()

print(f"Top performers avg units per transaction: {top_units:.1f}")
print(f"Bottom performers avg units per transaction: {bottom_units:.1f}")

# Check average price per unit
top_price_per_unit = df[df['State'].isin([' NSW', ' VIC'])]['Sales'].sum() / df[df['State'].isin([' NSW', ' VIC'])]['Unit'].sum()
bottom_price_per_unit = df[df['State'].isin([' NT', ' WA'])]['Sales'].sum() / df[df['State'].isin([' NT', ' WA'])]['Unit'].sum()

print(f"\nTop performers avg price per unit: ${top_price_per_unit:,.0f}")
print(f"Bottom performers avg price per unit: ${bottom_price_per_unit:,.0f}")
```

### Key Findings and Recommendations

Same price per unit: $2,500 everywhere
But top performers sell 4x more units per transaction: 33.4 vs 8.3 units
This means the revenue gap is purely about VOLUME per sale, not pricing.

1. Volume Discount Programs: "Buy 20+ units, get 10% off"
